```python
rr, EEz, EEr, HHt, EEz_GITC, EEr_GITC, HHt_GITC = ComputeModes("donneesCable", 0, 0.5, 0.0001)
```

```python
import matplotlib.pyplot as plt
```

```python
ind = 500
rEx = rr[ind]
rEx
```

```python
V = 40000
valEx = V/(rEx * np.log(0.092/0.0175))
coefGITC = valEx/EEr_GITC[ind]
coefSans = valEx/EEr[ind]
plt.semilogy(rr, abs(coefSans*EEr), 'r-')
plt.semilogy(rr, abs(coefGITC*EEr_GITC), 'b--')
plt.legend(["Avec blindage","Condition équivalente"])
```

```python
I = 900
valEx = I/(2*pi*rEx)
coefHGITC = valEx/HHt_GITC[ind]
coefHSans = valEx/HHt[ind]
plt.semilogy(rr, abs(coefHSans*HHt), 'r-')
plt.semilogy(rr, abs(coefHGITC*HHt_GITC), 'b--')
plt.legend(["Avec blindage","Condition équivalente"])
```

```python
plt.semilogy(rr, abs(coefHSans*EEz), 'r-')
plt.semilogy(rr, abs(coefHGITC*EEz_GITC), 'b--')
plt.legend(["Avec blindage","Condition équivalente"])
```
